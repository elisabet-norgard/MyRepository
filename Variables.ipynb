# Variables and variable types

## "Regular" variables

1. strings
1. integers
1. floats

## "Collection" variables

1. tuples
1. lists
1. sets
1. dictionaries

From https://www.w3schools.com:

- **Strings** are for storing text
- **Integers** are for storing whole numbers, positive or negative (such as 123 or -456), without decimals.
- **Floats** are for storing positive and negative numbers with a decimal point, like 35.3, or -2.34
- **Lists** are collections which are ordered and changeable. Allows duplicate members.
- **Tuples** are collections which are ordered and unchangeable. Allows duplicate members.
- **Sets** are collections which are unordered, "unchangeable", and unindexed. No duplicate members.
- **Dictionaries** are collections which are ordered and changeable. No duplicate members.

## Lists

```python
homes = ["Sioux Falls", "Hagerstown", "Marstons Mills", "Sogndal", "Los Angeles", "Aarhus"]
streetNos = [825, 13124, 19214, 22, 9, 900, 43]

print(homes)

# access a certain index 
print(homes[3])

# change a certain index 
homes[3] = "Hogwarts"
print(homes[3])
print(homes)

print(len(homes))
print(len(streetNos))
```

## Dictionaries

```python
my_pet = {'name': 'D2', 
          'species': 'cat',
          'likes': 'sleeping',
          'dislikes': 'loud noises',
          'age': 10}
```

```python
my_pet['age']
```

```python
items = ['grapes', 'skyr', ['salt', 'pepper', 'cumin'], 'rice']
number = [1, 2, 7, 3]
priority = ['high', 'high', 'low', 'high']
```

```python
shopping = {'items': items,
            'number': number,
            'priority': priority}
```

```python
shopping['items']
```

```python
shopping['items'][2][1]
```

```python
things = ['grapes', 'skyr', ['salt', 'pepper', 'cumin'], shopping]
```

```python
thingDict = {'shopping': shopping,
             'priority': priority,
             'things': things}
```

```python
thingDict['things']  # very deep data structures!
```

```python
thingDict['things'][3]
```

```python
thingDict['things'][3]['items']
```

```python
thingDict['things'][3]['items'][2]
```

```python
thingDict['things'][3]['items'][2][1]
```
